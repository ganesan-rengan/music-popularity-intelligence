```python
import pandas as pd
import numpy as np
import sweetviz as sv
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../data/processed/cleaned.csv')
print(f"✅ Loaded! {df.shape[0]} rows × {df.shape[1]} columns")
```

```python
# This one line analyzes the entire dataset automatically
# target_feat='Stream' tells sweetviz what we want to predict
report = sv.analyze(df, target_feat='Stream')

# Save the report as HTML inside reports folder
report.show_html('../reports/eda_report.html')

print("✅ EDA report saved to reports/eda_report.html")
```

```python
# Key stats sweetviz doesn't highlight clearly
print("── Dataset Shape ──────────────────")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\n── Stream Stats ───────────────────")
print(df['Stream'].describe())

print("\n── Album Type Counts ──────────────")
print(df['Album_type'].value_counts())

print("\n── Official Video Counts ──────────")
print(df['official_video'].value_counts())
```
